```python
def gerar_cracha(nome: str, idade: int, cargo: str):
  return (f"Crachá: {nome} {idade} anos - {cargo}")

gerar_cracha("Fulano", 32, "Developer")
```

```python
def enviar_email(destinatario, assunto="sem assunto"):
  return f"enviando e-mail para {destinatario} com o assunto: {assunto}"

enviar_email("legal@gmail.com")
#enviar_email("fulano@gmail.com", "Venda")
```

```python
def registrar_venda(id_venda,/,produto, *, desconto, impresso=False):

  return (f"Venda {id_venda} - Produto: {produto} | Desconto: {desconto} % | Recibo:{impresso}")

print(registrar_venda(1, "Pneu Pirelli", desconto=10))
print(registrar_venda(2, "AAA", desconto=10, impresso=True))
```

```python

def montar_perfil( **kwargs):
  print (" --- PERFIL USUARIO ---")
  for chave, valor in kwargs.items():
    print(f"{chave}: {valor}")
  return kwargs ## Retornar todas as características do usuario

montar_perfil(nome="Fulano", sobrenome="da Silva", idade=30, profissao="Programador")
```
