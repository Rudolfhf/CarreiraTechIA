```python
vogais = ["a", "e", "i", "o", "u"]
count = 0
def contador_vogais(palavra):
  global count
  for v in palavra:
    if v in vogais:
      count = count + 1
contador_vogais("abcde")
print(f"numero de vogais é {count}")
```

```python
def classificar_temperatura(temp):
  if not isinstance(temp, (int, float)):
        raise TypeError("A temperatura deve ser int ou float")
  if temp < 15:
      return print("frio")

  elif 15 < temp <=25: # elif temp >= 15 or temp <=25:
    return print("agradavel")

  else:
    return print("quente")

classificar_temperatura(9)
```

```python
def validador_senha(password):
  proibited_word = "senha"
  especial_caracter = ["#", "@", "*"]

  if len(password) >=8:
    print("passou de 8")
    if proibited_word not in password:
      print("Não há a palavra senha")
      if any(c in especial_caracter for c in password):
        print("Contém caractéres especiais")
        if any(c.isdigit() for c in password):
          print("Senha ok")


validador_senha("a#@*aafesenha")


```
